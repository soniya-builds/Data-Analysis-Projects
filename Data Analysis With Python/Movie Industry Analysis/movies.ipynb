```python
df = df.dropna()
```

```python
df.fillna(0, inplace=True)
```

Number of movies released per year

```python
movies_per_year = df['year'].value_counts().sort_index()

movies_per_year = df['year'].value_counts().sort_index()

plt.figure(figsize=(8,4))
movies_per_year.plot(kind='bar')
plt.title("Movies Released Per Year")
plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.show()

```

Top 10 most common genres

```python
top_genres = df['genre'].value_counts().head(10)

plt.figure(figsize=(8,4))
top_genres.plot(kind='bar')
plt.title("Top 10 Movie Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Movies")
plt.show()
```

```python
df.columns
```

Clean columns & convert numeric values

```python
# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Convert numeric columns safely
numeric_cols = ['score', 'budget', 'gross', 'runtime', 'votes']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Drop rows with missing important values
df = df.dropna(subset=['gross', 'score'])
```

Average movie score by genre

```python
avg_score_genre = (
    df.groupby('genre')['score']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(8,4))
avg_score_genre.plot(kind='bar')
plt.title("Average Movie Score by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Score")
plt.show()
```

Top 10 highest grossing movies

```python
top_gross_movies = df.sort_values(by='gross', ascending=False).head(10)

plt.figure(figsize=(8,4))
plt.bar(top_gross_movies['name'], top_gross_movies['gross'])
plt.title("Top 10 Highest Grossing Movies")
plt.xlabel("Movie Name")
plt.ylabel("Gross Revenue")
plt.xticks(rotation=90)
plt.show()
```

Average gross revenue by genre

```python
avg_gross_genre = (
    df.groupby('genre')['gross']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(8,4))
avg_gross_genre.plot(kind='bar')
plt.title("Average Gross Revenue by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Gross Revenue")
plt.show()
```

Average budget by genre

```python
avg_budget_genre = (
    df.groupby('genre')['budget']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(8,4))
avg_budget_genre.plot(kind='bar')
plt.title("Average Budget by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Budget")
plt.show()
```

Top 10 most voted movies

```python
top_voted = df.sort_values(by='votes', ascending=False).head(10)

plt.figure(figsize=(8,4))
plt.bar(top_voted['name'], top_voted['votes'])
plt.title("Top 10 Most Voted Movies")
plt.xlabel("Movie Name")
plt.ylabel("Votes")
plt.xticks(rotation=90)
plt.show()
```

Which countries produce the most movies?

```python
top_countries = df['country'].value_counts().head(10)

plt.figure(figsize=(8,4))
top_countries.plot(kind='bar')
plt.title("Top 10 Movie Producing Countries")
plt.xlabel("Country")
plt.ylabel("Number of Movies")
plt.show()
```
