Clean columns & convert numeric values

```python
# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Convert numeric columns safely
numeric_cols = ['score', 'budget', 'gross', 'runtime', 'votes']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Drop rows with missing important values
df = df.dropna(subset=['gross', 'score'])
```

Average movie score by genre

```python
avg_score_genre = (
    df.groupby('genre')['score']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(8,4))
avg_score_genre.plot(kind='bar')
plt.title("Average Movie Score by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Score")
plt.show()
```

Top 10 highest grossing movies

```python
top_gross_movies = df.sort_values(by='gross', ascending=False).head(10)

plt.figure(figsize=(8,4))
plt.bar(top_gross_movies['name'], top_gross_movies['gross'])
plt.title("Top 10 Highest Grossing Movies")
plt.xlabel("Movie Name")
plt.ylabel("Gross Revenue")
plt.xticks(rotation=90)
plt.show()
```

Do longer movies guarantee better audience scores?

```python
runtime_category = pd.cut(
    df['runtime'],
    bins=[0,90,120,180,300],
    labels=['Short','Medium','Long','Very Long']
)

avg_score_runtime = df.groupby(runtime_category)['score'].mean()

plt.figure(figsize=(8,4))
avg_score_runtime.plot(kind='bar')
plt.title("Average Score by Runtime Length")
plt.xlabel("Runtime Category")
plt.ylabel("Average Score")
plt.show()
```
